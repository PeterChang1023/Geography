# `Landcover5_only` 的 `gridx01`：改前 vs 改後

畫出原檔與改過的 `gridx01.nc`（地表類型），並標出改了哪些格子。風格沿用 `Boundary condition/bc_for_Wei/figures/gridx01.png`（同樣的色標：0–10）。

輸出：`../../All_exp_landcover_from_Wei/Landcover5_only/gridx01_compare.png`

```python
import numpy as np, xarray as xr
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap, BoundaryNorm
from matplotlib.patches import Rectangle, Patch
import cartopy.crs as ccrs

ROOT = "/Users/peterchang/Projects/Geography"
ORIG = f"{ROOT}/Boundary condition/bc_for_Wei/data/netcdf/gridx01.nc"
NEW  = f"{ROOT}/All_exp_landcover_from_Wei/Landcover5_only/gridx01.nc"
OUT  = f"{ROOT}/All_exp_landcover_from_Wei/Landcover5_only/gridx01_compare.png"

o = xr.open_dataset(ORIG, decode_times=False); n = xr.open_dataset(NEW, decode_times=False)
go, gn = o.GRIDX01.values[0], n.GRIDX01.values[0]
lon = o.lon.values; lat = o.lat.values; latb = o.lat_bnds.values
changed = go != gn
print("改變的格子數：", int(changed.sum()))
for k, v in zip(*np.unique(go[changed], return_counts=True)):
    print(f"  原本 {int(k):>2} → 5 : {int(v)} 格")

# 格子邊界（lon 轉 -180..180 並排序，lat 邊界由 lat_bnds 來）
lon180 = np.where(lon > 180, lon - 360, lon); order = np.argsort(lon180)
xe = np.append(lon180[order] - 1.40625, lon180[order][-1] + 1.40625)
ye = np.append(latb[:, 0], latb[-1, 1])
R = lambda a: a[:, order]

names = {0: "sea", 1: "continental ice", 2: "broadleaf evergreen forest", 3: "broadleaf deciduous forest",
         4: "mixed forest", 5: "coniferous forest", 6: "high-lat deciduous forest", 7: "wooded C4 grassland",
         8: "shrubs & bare ground", 9: "tundra", 10: "C3 grassland"}
cols = list(plt.cm.tab10.colors) + ["#9edae5"]                        # 與原圖相同的 11 色
cmap = ListedColormap(cols); norm = BoundaryNorm(np.arange(-0.5, 11.5, 1), 11)

def coast(ax):
    try:
        ax.coastlines(lw=0.5)
    except Exception as e:
        print("沒有海岸線：", type(e).__name__)
```

## 圖：改前 / 改後 / 改了哪裡

```python
fig = plt.figure(figsize=(17, 12), dpi=130)
proj = ccrs.Robinson(); pc = ccrs.PlateCarree()
ext = (20, 140, 15, 75)

def draw(ax, field, title, zoom=False, mark=None):
    ax.pcolormesh(xe, ye, R(field), cmap=cmap, norm=norm, transform=pc)
    if mark is not None:
        for j, i in zip(*np.where(mark)):
            x0 = lon180[i] - 1.40625
            ax.add_patch(Rectangle((x0, latb[j, 1]), 2.8125, latb[j, 0] - latb[j, 1], fill=False, ec="red", lw=1.4, transform=pc, zorder=5))
    coast(ax)
    if zoom:
        ax.set_extent(ext, crs=pc)
        gl = ax.gridlines(draw_labels=True, lw=0.4, ls=":", color="0.4"); gl.top_labels = gl.right_labels = False
    else:
        ax.set_global()
    ax.set_title(title, fontsize=11, loc="left", fontweight="bold")

ax1 = fig.add_subplot(2, 2, 1, projection=proj); draw(ax1, go, "A. ORIGINAL gridx01 (global)")
ax2 = fig.add_subplot(2, 2, 2, projection=proj); draw(ax2, gn, "B. MODIFIED gridx01 (global) - Landcover5_only")
ax3 = fig.add_subplot(2, 2, 3, projection=pc);   draw(ax3, go, "C. ORIGINAL, zoom on Asia", zoom=True)
ax4 = fig.add_subplot(2, 2, 4, projection=pc);   draw(ax4, gn, f"D. MODIFIED, zoom: red boxes = {int(changed.sum())} cells whose value changed", zoom=True, mark=changed)

# 全球圖上也標出改的範圍
for ax in (ax2,):
    for j, i in zip(*np.where(changed)):
        ax.add_patch(Rectangle((lon180[i] - 1.40625, latb[j, 1]), 2.8125, latb[j, 0] - latb[j, 1], fill=False, ec="red", lw=0.6, transform=pc, zorder=5))

handles = [Patch(fc=cols[k], ec="0.3", label=f"{k}  {names[k]}") for k in range(11)]
fig.legend(handles=handles, loc="lower center", ncol=4, fontsize=9, frameon=False, bbox_to_anchor=(0.5, 0.0))
fig.suptitle("gridx01 (primary land-cover class): 110 T42 cells in the FRONT source region (inside red line, outside DES) set to 5; "
             "all other cells unchanged", fontsize=12, y=0.98)
fig.tight_layout(rect=(0, 0.07, 1, 0.96))
fig.savefig(OUT, bbox_inches="tight")
print("已存", OUT)
```
